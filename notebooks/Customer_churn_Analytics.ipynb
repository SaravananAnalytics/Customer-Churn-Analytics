```python
df.columns = df.columns.str.strip().str.lower()
```

```python
df["totalcharges"] = pd.to_numeric(df["totalcharges"], errors="coerce")
```

```python
print(df[df["totalcharges"].isnull()][["customerid", "tenure"]])
```

```python
df["totalcharges"] = df["totalcharges"].fillna(0)
```

```python
df.isnull().sum()
```

```python
print(df["customerid"].duplicated().sum())
```

```python
svc = ["multiplelines", "onlinesecurity", "onlinebackup", "deviceprotection",
"techsupport", "streamingtv", "streamingmovies"]
df[svc] = df[svc].replace({"No internet service": "No", "No phone service": "No"})
df["seniorcitizen"] = df["seniorcitizen"].map({0: "No", 1: "Yes"})
df["paymentmethod"] = df["paymentmethod"].str.replace(" (automatic)", "", regex=False)
```

```python
df["churnflag"] = (df["churn"] == "Yes").astype(int)
bins = [-1, 12, 24, 48, 72]
labels = ["0-12 months", "13-24 months", "25-48 months", "49-72 months"]
df["tenuregroup"] = pd.cut(df["tenure"], bins=bins, labels=labels)
df["tenureorder"] = df["tenuregroup"].cat.codes + 1
df["tenuregroup"] = df["tenuregroup"].astype(str)
```
